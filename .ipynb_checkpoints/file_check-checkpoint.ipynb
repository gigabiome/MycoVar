```python
# import functions and other modules
import functions
import importlib
importlib.reload(functions)
import os
import pandas as pd
import warnings
warnings.simplefilter(action='ignore', category=UserWarning)
warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)

# Global variables for files
INPUT = 'input/'
OUTPUT = 'output/'
BIN = 'bin/'
```

```python
def checkSNPFiles(snp_files,cov_files,stat_files):
    remove_samples = []
    for file in snp_files:
        df = pd.read_csv(INPUT+'snp/'+file)
        if 'Reference Position' not in df.columns:
            remove_samples.append(file.replace('.snp',''))
            print('❌File '+file+' is not a SNP file')
    return remove_samples

def fileOverlap(snp_files,cov_files,stat_files):
    from collections import Counter
    snp_files = [x.split('.')[0] for x in snp_files]
    cov_files = [x.split('.')[0] for x in cov_files]
    stat_files = [x.split('.')[0] for x in stat_files]
    
    files = snp_files + cov_files + stat_files
    
    counted = Counter(files)
    not_matched = []
    for f in counted:
        if counted[f] != 3:
            not_matched.append(f)
    return not_matched
```

```python
# Setup input
functions.setupInput(INPUT)
snp_files = os.listdir(INPUT+'snp/')
cov_files = os.listdir(INPUT+'cov/')
stat_files = os.listdir(INPUT+'stat/')
```

```python
check_samples = checkSNPFiles(snp_files,cov_files,stat_files)
```

```python
for file in check_samples:
    cmd = 'mv '+INPUT+'snp/'+file+'.snp '+INPUT+'check_sample/'
    #os.system(cmd)
    print(cmd)
    
    cmd = 'mv '+INPUT+'stat/'+file+'.stat '+INPUT+'check_sample/'
    #os.system(cmd)
    print(cmd)
    
    cmd = 'mv '+INPUT+'cov/'+file+'.cov '+INPUT+'check_sample/'
    #os.system(cmd)
    print(cmd)
```

```python
files = fileOverlap(snp_files,cov_files,stat_files)

for file in files:
    cmd = 'mv '+INPUT+'snp/'+file+'.snp '+INPUT+'check_sample/'
    os.system(cmd)
    print(cmd)
    
    cmd = 'mv '+INPUT+'stat/'+file+'.stat '+INPUT+'check_sample/'
    os.system(cmd)
    print(cmd)
    
    cmd = 'mv '+INPUT+'cov/'+file+'.cov '+INPUT+'check_sample/'
    os.system(cmd)
    print(cmd)
```
