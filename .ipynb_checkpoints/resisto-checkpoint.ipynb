# Resisto

This notebook is used to analyse SNP files and perform resistotyping generating several output files

1. Set global parameters for the run that will be used throughout the script
2. Run the setup function to setup the input and output
2. Generate output

```python
# Parameters for snp calling (quality, forward/reverse balance, frequency, coverage)
QUAL = 20
FRB = 0
FREQ = 5
COV = 5
```

```python
# Setup input
functions.setupInput(INPUT)
snp_files = os.listdir(INPUT+'snp/')
cov_files = os.listdir(INPUT+'cov/')
stat_files = os.listdir(INPUT+'stat/')

# Setup output. Note: it will clear everything in the resisto output
functions.setupOutput(OUTPUT,'resisto')

# Parse databases for resistotyping
DR_WHO,DR_CTB,DR_GENES,dCTB,dWHO,dGene2Drug,dDrug2Gene,DR_WHO_promotors = functions.setupResisto(BIN)
```

```python
# perform resistotyping
params = [QUAL,FRB,FREQ,COV,INPUT,OUTPUT,BIN]
resisto_list = [DR_WHO,DR_CTB,DR_GENES,dCTB,dWHO,dGene2Drug,dDrug2Gene,DR_WHO_promotors]

# select here the stats you want to pull into the resulting files
# this can be any column from the snp, cov or stat file
# NB: do not add frequency here because it will be written out by default
stats = ['Coverage','Zero coverage bases','Mapped reads','Gene deletion proportion']

out = functions.mainResistotyping(snp_files,params,resisto_list,stats)
```

```python
d = {'AWR':'#FF0000',
     'AWR-I':'#FF0000',
     'AWRI-ER':'#FF0000',
     'US':'#808080',
     'NAWR':'#008000',
     'NAWR-I':'#008000',
     'CTB-R':'#FF0000',
     'UNGR':'#0000FF',
     'WT':'#FFFFFF'}

for drug in out.columns:
    fout = open('output/resisto/iTol/'+drug+'.txt','w')
    fout.write('DATASET_COLORSTRIP\nSEPARATOR TAB\nDATASET_LABEL\t'+drug+'\nCOLOR\t#ff0000\n')
    fout.write('\nDATA\n')
    df = out[[drug]]
    for index,row in df.iterrows():
        fout.write(index+'\t')
        fout.write(d[row[drug]] + '\t')
        fout.write(row[drug])
        fout.write('\n')
    fout.close()
        
```
